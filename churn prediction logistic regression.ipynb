# keeping important columns

# label encoding

```python
from sklearn.preprocessing import LabelEncoder
label_encoder=LabelEncoder()
```

# but you want to do all at once

```python
# using python
```

```python
# Correct categorical list — remove numeric columns
categorical_cols = ['gender','SeniorCitizen','Partner','Dependents',
                     'PhoneService','MultipleLines','Contract','Churn']

for col in categorical_cols:
    df[col] = label_encoder.fit_transform(df[col])

# Make sure TotalCharges stays numeric (sometimes it loads as string/object)
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

# tenure also stays as-is (numeric), no encoding needed
```

```python
df
```

# jugaad use binary and mapping replace 

```python
df2=df.copy()
binary_cols= ['Partner','Dependents','Churn']# when colms has only Yes and No

df2[binary_cols]= df2[binary_cols].replace({'Yes':1,'No':0})
```

```python
df2
```

```python
df3=df.copy()
df3['gender']=df3['gender'].replace({'Female' :0, "Male" :1})
```

```python
df # using python loop
```

# remeber that regeression does not take null values and when train and test split is done, it capture the data at that moment and if train_test is ran then nan values can not be removed after that

```python
df['TotalCharges']=df['TotalCharges'].fillna(df['TotalCharges'].mean(),inplace=True)
```

```python
# split data train and test
```

```python
x=df.drop('Churn',axis=1)
y=df['Churn']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train.head()
```

```python
x_train.info()
```

```python
x_train['TotalCharges']=pd.to_numeric(x_train['TotalCharges'],errors='coerce')
x_test['TotalCharges']=pd.to_numeric(x_test['TotalCharges'],errors='coerce')
```

```python
x_train.isnull().sum()
```

```python
x_train['TotalCharges'].fillna(x_train['TotalCharges'].mean(),inplace=True)
x_test['TotalCharges'].fillna(x_test['TotalCharges'].mean(),inplace=True)
```

```python
# stadardize features (optinal in logistic regression)
```

```python

from sklearn.preprocessing import StandardScaler
scalar= StandardScaler()
```

```python
x_train=scalar.fit_transform(x_train)
x_test=scalar.transform(x_test)
```

```python
x_train
```

```python
print(df['TotalCharges'].isnull().sum()) # reomve null values as it can hinder machine learing in Regression as regression doesn't accept null values
```

```python
print(df['TotalCharges'].isnull().sum())
```

```python
from sklearn.linear_model import LogisticRegression
lg = LogisticRegression()

```

```python
print(df.isnull().sum())          # check on the full dataframe before split
print(np.isnan(x_train).sum())    # if x_train is a numpy array
```

```python
lg.fit(x_train,y_train)
y_pred=lg.predict(x_test)
```

```python
y_pred
```

```python
# accuracy score
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,y_pred)
```

```python
gender = 'female'
SeniorCitizen = 'No'
Partner = 'Yes'
Dependents = 'No'
tenure = 1
PhoneService = 'No'
MultipleLines = 'No Phone Service'
Contract = 'month-to-month'
TotalCharges = 25.05

# first make the variables before giving them into dictionary
```

```python
data = {
    'gender': [gender],
    'SeniorCitizen': [SeniorCitizen],
    'Partner': [Partner],
    'Dependents': [Dependents],
    'tenure': [tenure],
    'PhoneService': [PhoneService],
    'MultipleLines': [MultipleLines],
    'Contract': [Contract],
    'TotalCharges': [TotalCharges]
}

df1 = pd.DataFrame(data)
```

```python
df1# this is individual detail we encode it to get our answer as 0(no churn) or 1 (churn)
```

```python
def predictive(gender,SeniorCitizen,Partner,Dependents,tenure,PhoneService,MultipleLines,Contract,TotalCharges):# we created a function and called it based on our features 
    data={
        'gender':[gender],	# 'gender': is colm name and [gender]='female' (we are passing value in it)
        'SeniorCitizen':[SeniorCitizen],	
        'Partner':[Partner],
        'Dependents':[Dependents],
        'tenure':[tenure],
        'PhoneService':[PhoneService],
        'MultipleLines':[MultipleLines],	
        'Contract':	[Contract],
        'TotalCharges':[TotalCharges]
    }
    df1=pd.DataFrame(data)

    categorical_colms=['gender','SeniorCitizen','Partner','Dependents','tenure','PhoneService','MultipleLines','Contract','TotalCharges']
    for colms in categorical_colms:
        df1[colms]= label_encoder.fit_transform(df1[colms])
    

    df1=scalar.fit_transform(df1)
    result=lg.predict(df1).reshape(1,-1)
    return result[0]
```
